```python
import csv

# Read the CSV file
a = []

with open("weather dataset.csv", "r") as csvfile:
    reader = csv.reader(csvfile)

    for row in reader:
        if row:  # Ignore empty rows
            a.append(row)

# Display the data
print("Training Data:")
for row in a:
    print(row)

# Check whether data was loaded
if len(a) == 0:
    print("\nERROR: No data was found in the CSV file.")
else:
    print("\nThe total number of training instances are:", len(a))

    # Number of attributes excluding class/target
    num_attribute = len(a[0]) - 1

    # Initial hypothesis
    hypothesis = ["0"] * num_attribute

    print("\nThe initial hypothesis is:")
    print(hypothesis)

    # FIND-S algorithm
    for i in range(len(a)):

        # Check positive example
        if a[i][num_attribute].lower() == "yes":

            for j in range(num_attribute):

                if hypothesis[j] == "0":
                    hypothesis[j] = a[i][j]

                elif hypothesis[j] != a[i][j]:
                    hypothesis[j] = "?"

            print("\nThe hypothesis for training instance", i + 1, "is:")
            print(hypothesis)

    print("\nThe Maximally Specific Hypothesis is:")
    print(hypothesis)
```
